```python
import numpy as np

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

w = 0.0
lr = 0.01
epochs = 100

for epoch in range(epochs):
    for i in range(len(X)):
        y_pred = w * X[i] #single data point prediction
        gradient = 2 * (y_pred - y[i]) * X[i]
        w = w - lr * gradient

print("Final weight:", w)
```
