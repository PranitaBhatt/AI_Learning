```python
import numpy as np

# Data
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Parameters
w = 0.0
lr = 0.01
epochs = 100

n = len(X)

for epoch in range(epochs):
    y_pred = w * X
    gradient = (2/n) * np.sum((y_pred - y) * X)
    w = w - lr * gradient

print("Final weight:", w)
```
