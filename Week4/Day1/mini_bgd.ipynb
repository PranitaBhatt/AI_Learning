```python
import numpy as np

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

w = 0.0
lr = 0.01
epochs = 100
batch_size = 2

n = len(X)

for epoch in range(epochs):
    for i in range(0, n, batch_size):
        X_batch = X[i:i+batch_size]#updating the batch of data points
        y_batch = y[i:i+batch_size]

        y_pred = w * X_batch
        gradient = (2/len(X_batch)) * np.sum((y_pred - y_batch) * X_batch)
        w = w - lr * gradient

print("Final weight:", w)
```
