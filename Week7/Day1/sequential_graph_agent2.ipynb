````python
from typing import TypedDict, List
from langgraph.graph import StateGraph
````

````python
class AgentState(TypedDict):
    name:str
    age:int
    final_result:str
    skills:List[str]
````

````python
def first_node(state: AgentState) -> dict:
    """This node will used to greet the user"""
    return {"final_result": f"{state['name']} welcome to the system!"}

def second_node(state: AgentState) -> dict:
    """Describes the user age"""
    return {"final_result": state["final_result"] + f" You are {state['age']} years old."}

def third_node(state: AgentState) -> dict:
    """This node will list the user's skills in a formatted string"""
    skills = ", ".join(state["skills"])
    return {"final_result": state["final_result"] + f" You have skills in: {skills}."}
````

````python
graph=StateGraph(AgentState)
graph.add_node("Introduction",first_node)
graph.add_node("Age",second_node)
graph.add_node("Skills",third_node)

graph.set_entry_point("Introduction")
graph.add_edge("Introduction","Age")
graph.add_edge("Age","Skills")
graph.set_finish_point("Skills")
app=graph.compile()
````

````python
#Visualizing this agent code
from IPython.display import Markdown, display
display(Markdown(f"```mermaid\n{app.get_graph().draw_mermaid()}\n```"))
````

````python
result = app.invoke({"name": "Linda", "age": 31, "skills": ["Python", "Machine Learning", "LangGraph"]})
print(result["final_result"])
````
