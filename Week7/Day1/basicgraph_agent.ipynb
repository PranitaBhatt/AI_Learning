````python
from typing import Dict,TypedDict
from langgraph.graph import StateGraph
````

````python
#creating a agent state
#State SCHEMA
class AgentState(TypedDict): #state needs to be in form of TypedDict
    message: str #the message passed will be a string

#first node
def greet_node(state:AgentState)->AgentState:  #so input state is passed inside the () and -> this represents the o/p state
    """ Simple node that adds a greeting message to the state""" #passing a DocString
    state['message']="Hey" + state["message"]+",How is your day going?" #will update the state
    return state
````

````python
#Building a Graph
graph=StateGraph(AgentState) #PASSING STATE SCHEMA HERE
graph.add_node("greeter",greet_node) #GREET NODE

#STARTING POINT
graph.set_entry_point("greeter")
#END POINT
graph.set_finish_point("greeter")

#we need to COMPILE this graph
app=graph.compile()
````

````python
#Visualizing this agent code
from IPython.display import Markdown, display

display(Markdown(f"```mermaid\n{app.get_graph().draw_mermaid()}\n```"))
````

````python
result=app.invoke({'message':'Swara'})
````
