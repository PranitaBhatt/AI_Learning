```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import json
```

```python
llm = ChatOllama(
    model="llama3.1:8b",
    temperature=0.7
)
```

```python
topic = input("Enter a technical topic: ")
```

```python
prompt = ChatPromptTemplate.from_template("""
You are an expert AI technical educator.

For the topic: {topic}

Generate the following:

1. Technical Explanation
   - Beginner friendly
   - Technical but easy to understand
   - Include examples

2. Architecture Summary
   - Main components
   - Workflow
   - Technologies used

3. Simple Flow Diagram

Keep the explanation detailed.
""")
```

```python
chain = (
    prompt
    | llm
    | StrOutputParser()
)

```

```python
result = chain.invoke({
    "topic": topic
})
```

```python
print("\n")
print("=" * 60)
print("AI DOCUMENTATION ASSISTANT OUTPUT")
print("=" * 60)
print(result)
```

```python
output_data = {
    "topic": topic,
    "generated_content": result
}

with open("ai_documentation_output.json", "w") as file:
    json.dump(output_data, file, indent=4)

print("\n")
print("Structured JSON exported successfully!")
print("Saved as: ai_documentation_output.json")
```
